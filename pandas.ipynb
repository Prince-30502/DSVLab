```python


sales = pd.Series({
    'Laptop': 800,
    'Mobile': 600,
    'Mouse': 250,
    'Keyboard': 450
})

print("Sales Series:")
print(sales)

print("\nValues:", sales.values)
print("Index:", sales.index)

print("\nLaptop Sales:", sales['Laptop'])

print("\nSales Above 500:")
print(sales[sales > 500])

print("\nDoes Laptop Exist?", 'Laptop' in sales.index)

print("\nDictionary:")
print(sales.to_dict())
```
