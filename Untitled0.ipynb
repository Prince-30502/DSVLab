```python
course_df = pd.DataFrame({
    'Student': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Course': ['Python', 'Java', 'C++', 'Python', 'Java'],
    'Score': [85, 90, 78, 88, 92]
})

print("\nQ26 - Column names:")
print(course_df.columns)

print("\nQ26 - Student column:")
print(course_df['Student'])

course_new_df = course_df[['Student', 'Score']].copy()
course_new_df['Batch'] = '2026'
print("\nQ26 - Updated DataFrame:")
print(course_new_df)
```

```python
# Q28. Create school_df and request missing House column
school_df = pd.DataFrame({
    'Student': ['Aman', 'Riya', 'Rahul'],
    'Class': [10, 11, 12],
    'Section': ['A', 'B', 'C']
})

school_new_df = school_df.reindex(
    columns=['Student', 'Class', 'Section', 'House']
)

school_new_df['House'] = 'Green House'
print("\nQ28:")
print(school_new_df)
```
